## Задание 1. Метод Гаусса 
Решить систему 4-го порядка методом Гаусса с выбором и без выбора главного элемента с точностью $\epsilon = 0.001$.
Сравнить, привести пример, где метод без выбора дает сбой.

```python
A1 = np.array(
    [
        [2.20, -3.17, 1.24, -0.87],
        [1.50, 2.11, -0.45, 1.44],
        [0.86, -1.44, 0.62, 0.28],
        [0.48, 1.25, -0.63, -0.97]
    ]
)

b1 = np.array([0.46, 1.50, -0.12, 0.35])
```

### Метод Гаусса

```python
def gauss(A_in, b_in, piv_point):
    A = A_in.copy()
    b = b_in.copy()
    n = len(b)
    for k in range(n - 1):
        # выбор главного элемента
        if piv_point:
            max_row = np.argmax(np.abs(A[k:, k])) + k

            if max_row != k:
                A[[k, max_row]] = A[[max_row, k]]
                b[[k, max_row]] = b[[max_row, k]]

        for i in range(k + 1, n):

            factor = A[i, k] / A[k, k]

            A[i, k:] = A[i, k:] - factor * A[k, k:]

            b[i] = b[i] - factor * b[k]
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        x[i] = (b[i] - np.dot(A[i, i + 1 :], x[i + 1 :])) / A[i, i]
    return x
```

Решить систему 4-го порядка с точностью $\epsilon = 0.0001$.
Система задана в каноническом виде $X = \beta + \alpha X$.

```python
A2 = np.array(
    [
        [0.00, 0.22, -0.11, 0.31],
        [0.38, 0.00, -0.12, 0.22],
        [0.11, 0.23, 0.00, -0.51],
        [0.17, -0.21, 0.31, 0.00]
    ]
)
B2 = np.array([2.70, -1.50, 1.20, -0.17])
```

## Задание 3. Обратная матрица методом Гаусса-Жордана
Решить систему 3-го порядка. <br>
Метод заключается в приписывании единичной матрицы справа от основной $[A | E]$ и выполнении элементарных преобразований строк до получения $[E | A^{-1}]$.

```python
A3 = np.array([
    [2, 1, 1],
    [1, 3, 2],
    [4, 1, 2]
], dtype=float)
b3 = np.array([5, 8, 9], dtype=float)
```

### Получение обратной матрицы

```python
# - [x] Расписать получение обратной матрицы ✅ 2026-03-27

def get_inverse_gauss_jordan(A_in):
    n = A_in.shape[0] #определение размерности матрицы
    E = np.eye(n) #создание единичной матрицы размерности n x n

    AE = np.hstack((A_in, E)) #создание объединённой матрицы из исходной и единичной A|E
    #если преобразовать левую часть к единичной матрице, правая часть станет обратной

    #цикл приведения левой матрицы к единичной
    for i in range(n):
        max_row = np.argmax(np.abs(AE[i:, i])) + i # находим максимальный i i элемент (опорный элемент)
        if AE[max_row, i] == 0: #если главный элемент равен нулю - матрица необратима
            raise ValueError("Обратной матрицы не существует")

        # Перестановка строк
        if max_row != i:
            AE[[i, max_row]] = AE[[max_row, i]]

        pivot = AE[i, i] #на что делить чтобы получить 1 на диагонали
        AE[i] = AE[i] / pivot #делим чтобы получить 1

        # элементы к нулям в столбце i выше и ниже диагонали
        for j in range(n):
            if j != i:
                factor = AE[j, i]
                AE[j] -= factor * AE[i]

    # обратная матрица получена.
    A_inv = AE[:, n:]
    return A_inv
```

### Ответы

```python
A3_inv = get_inverse_gauss_jordan(A3)
print("Обратная матрица A^-1:")
print(A3_inv)

X_inv_method = np.dot(A3_inv, b3)
print(f"\nРешение системы: {X_inv_method}")

c3 = nevyazka(A3, X_inv_method, b3)
print(f"Невязка: {c3}")
```

## Задание 4 Метод Ньютона
Решить систему: <br>
$f_1(x, y) = x - e^{0.1-y} + 2.1 = 0$ <br>
$f_2(x, y) = y + \cos(x) = 0$ <br>
С точностью $\epsilon = 0.001$.<br>

Метод Ньютона для систем: $X^{(k+1)} = X^{(k)} - J^{-1}(X^{(k)}) \cdot F(X^{(k)})$, где $J$ - матрица Якоби. <br>

Матрица Якоби: <br>
$J = \begin{pmatrix}
\frac{\partial f_1}{\partial x} & \frac{\partial f_1}{\partial y} \\
\frac{\partial f_2}{\partial x} & \frac{\partial f_2}{\partial y}
\end{pmatrix} = \begin{pmatrix}
1 & -e^{0.1-y} \\ 
-\sin(x) & 1
\end{pmatrix}$

### График

```python
x_range = np.linspace(-5, 5, 100)
y_range = np.linspace(-5, 2, 100)
X, Y = np.meshgrid(x_range, y_range)

F1 = X + np.exp(0.1 - Y) + 2.1
F2 = Y + np.cos(X)

plt.figure(figsize=(10, 8))

contour1 = plt.contour(X, Y, F1, levels=[0], colors="blue", linewidths=2)
contour2 = plt.contour(X, Y, F2, levels=[0], colors="red", linewidths=2)

plt.xlabel("x", fontsize=12)
plt.ylabel("y", fontsize=12)
plt.grid(True, alpha=0.3)

plt.show()
```

```python
# - [x] Расписать формулу ньютона ✅ 2026-03-27
#функция, возвращающая значения функций в точке X
def F_nonlin(X):
    x, y = X[0], X[1] # x,y из точки X
    return np.array([
        x + np.exp(0.1 - y) + 2.1, # значение f1
        y + np.cos(x)               # значение f2
                    ])

#функция вычисления якобиана
def Jacobian(X):
    x, y = X[0], X[1] # Координаты точки
    return np.array([#возвращаем матрицу
        [1, -np.exp(0.1 - y)],
        [-np.sin(x), 1]
                     ])



X_nl = np.array([-3.0, -2.0]) #начальное приближение
eps_nl = 0.001 #допустимая погрешность
iter_nl = 0 #счётчик

while True:
    iter_nl += 1
    F_val = F_nonlin(X_nl) # значения функций в точке X_nl
    J_val = Jacobian(X_nl) # матрица Якоби в точке X_nl

    # J · dX = -F
    # v
    dX = np.linalg.solve(J_val, -F_val)

    X_next = X_nl + dX

    if np.max(np.abs(dX)) <= eps_nl: # погрешность достигнута
        X_nl = X_next
        break
    X_nl = X_next # новое приближение

print(f"Корень найден за {iter_nl} итераций.")
print(f"x = {X_nl[0]:.5f}, y = {X_nl[1]:.5f}")
```

## Задание 6. LU-разложение матрицы
Выполнить LU-разложение для матрицы из Задания 1: $P \cdot A = L \cdot U$. <br>
Решить $Ax = b$ для 100 случайных векторов свободных членов, используя полученные $L$ и $U$. <br>
Уравнение $A \cdot x = b$ преобразуется в:
1) $P^-1 \cdot L \cdot U \cdot x = b \implies L \cdot (U \cdot x) = P \cdot b$
2) Решаем $L \cdot y = P \cdot b$ (прямой ход)
3) Решаем $U \cdot x = y$ (обратный ход)

```python
# перестановки P, нижнюю L и верхнюю U
P, L, U = la.lu(A1)

print("Матрица L:\n", np.round(L, 3))
print("Матрица U:\n", np.round(U, 3))


# Функция решения Ly = Pb, затем Ux = y
def solve_lu(P, L, U, b_vec):
    n = len(b_vec)
    # Умножаем матрицу перестановок на вектор b
    Pb = np.dot(P, b_vec)

    # Прямая подстановка L * y = Pb
    y = np.zeros(n)
    for i in range(n):
        y[i] = Pb[i] - sum(L[i, j] * y[j] for j in range(i))

    # Обратная подстановка U * x = y
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        x[i] = (y[i] - sum(U[i, j] * x[j] for j in range(i + 1, n))) / U[i, i]

    return x


# Решение для 100 случайных векторов
num_vectors = 100
random_bs = np.random.rand(num_vectors, 4)
results_lu = []

start_time = time.time()
for i in range(num_vectors):
    sol = solve_lu(P, L, U, random_bs[i])
    results_lu.append(sol)
time_lu_100 = time.time() - start_time

print(f"\nВремя решения для 100 случайных векторов: {time_lu_100:.5f} сек.")
```

Использование заранее вычисленных матриц L и U позволяет многократно решать систему с разными правыми частями за очень короткое время, так как этап исключения выполняется только 1 раз.
